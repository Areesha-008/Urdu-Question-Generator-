```python
from pathlib import Path
import subprocess
import sys

if 'google.colab' in sys.modules:
    from google.colab import drive
    drive.mount('/content/drive')
    ROOT = Path('/content/urdu-question-generator')
    if not ROOT.exists():
        subprocess.run(['git', 'clone', 'https://github.com/Areesha-008/Urdu-Question-Generator-.git', str(ROOT)], check=True)
    BASE = Path('/content/drive/MyDrive/Urdu-QG-v2')
else:
    ROOT = Path.cwd() if (Path.cwd() / 'config.py').exists() else Path.cwd().parent
    BASE = ROOT / 'artifacts'
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(ROOT / 'requirements.txt')], check=True)
sys.path.insert(0, str(ROOT))
DATA = BASE / 'data'
RUN = BASE / 'answer_gru_v1'
RUN.mkdir(parents=True, exist_ok=True)
print('Data:', DATA, 'Run:', RUN)
```

```python
from scripts.prepare_data import prepare, read_pairs

statistics = prepare(DATA)
print(statistics)
```

```python
import matplotlib.pyplot as plt

figure, axes = plt.subplots(1, 2, figsize=(10, 4))
for name in statistics:
    pairs = read_pairs(DATA / f'{name}.tsv')
    for index, label in enumerate(['Source words', 'Target words']):
        axes[index].hist([len(pair[index].split()) for pair in pairs], bins=25,
            alpha=.4, density=True, label=name)
        axes[index].set_xlabel(label)
        axes[index].legend()
figure.tight_layout()
figure.savefig(DATA / 'lengths.png', dpi=150)
```
